### Figure S5: Conceptual GDM spline interpretation

<figure>
<img src="attachment:images/gdm_conceptual.jpg" alt="Conceptual representation of generalized dissimilarity modelling (GDM). GDM relates observed compositional dissimilarity between pairs of assemblages to environmental and geographic differences through nonlinear I-spline transformations of predictor variables. The absolute differences in transformed predictor values between assemblages i and j are summed with a model intercept to estimate ecological distance (η). A negative exponential link function then relates ecological distance to predicted compositional dissimilarity, allowing turnover to increase nonlinearly and approach saturation. Reproduced from Figure 2 of Mokany et al. (2022), Global Ecology and Biogeography, doi:10.1111/geb.13459." />
<figcaption><strong>Conceptual representation of generalized dissimilarity modelling (GDM).</strong> GDM relates observed compositional dissimilarity between pairs of assemblages to environmental and geographic differences through nonlinear I-spline transformations of predictor variables. The absolute differences in transformed predictor values between assemblages i and j are summed with a model intercept to estimate ecological distance (η). A negative exponential link function then relates ecological distance to predicted compositional dissimilarity, allowing turnover to increase nonlinearly and approach saturation. Reproduced from Figure 2 of Mokany et al. (2022), Global Ecology and Biogeography, doi:10.1111/geb.13459.</figcaption>
